```python
import pandas as pd
import numpy as np
```

```python
RAW_DIR = "../data/raw"
PROCESSED_DIR = "../data/processed"
```

```python
huy = pd.read_csv(f"{RAW_DIR}/huy.csv")
ngan = pd.read_csv(f"{RAW_DIR}/ngannguyen.csv")
choe = pd.read_csv(f"{RAW_DIR}/choe.csv")
```

```python
print(huy['id'].duplicated().sum())
print(ngan['id'].duplicated().sum())
print(choe['id'].duplicated().sum())
```

```python
huy = huy.rename(columns={'label': 'label_huy'})
ngan = ngan.rename(columns={'label': 'label_ngan'})
choe = choe.rename(columns={'label': 'label_choe'})
```

```python
def normalize_label(df, column):
    df[column] = (
        df[column]
        .str.lower()
        .str.strip()
    )
    return df
```

```python
huy = normalize_label(huy, 'label_huy')
ngan = normalize_label(ngan, 'label_ngan')
choe = normalize_label(choe, 'label_choe')
```

```python
merged = pd.merge(
    huy,
    ngan[['id', 'data', 'label_ngan']],
    on=['id', 'data'],
    how='inner'
)
```

```python
merged = pd.merge(
    merged,
    choe[['id', 'data', 'label_choe']],
    on=['id', 'data'],
    how='inner'
)
```

```python
print(merged.head())
print(merged.shape)
```

```python
merged.to_csv(
    f"{PROCESSED_DIR}/merged_annotations.csv",
    index=False
)
```
